# Project configuration

Load the local dotenv settings and inspect the shared paths used by the workflow notebooks. Values such as output and source-reference paths are shown, but secrets from `params.env` are never printed.

```python
# ruff: noqa: E402
from pathlib import Path
import sys

CURRENT_DIRECTORY = Path.cwd().resolve()
CANDIDATE_ROOT = (
    CURRENT_DIRECTORY if (CURRENT_DIRECTORY / "app").is_dir() else CURRENT_DIRECTORY.parent
)
sys.path.insert(0, str(CANDIDATE_ROOT))
import app
from app.dotenv import (
    OUTPUT_ROOT,
    SOURCE_REFERENCE,
    data_dir,
    env_dir,
    mplconfig_dir,
    project_dir,
)

APP_PATH = Path(app.__file__).resolve()
print(f"Kernel executable: {sys.executable}")
print(f"Working directory: {CURRENT_DIRECTORY}")
print(f"Imported app package: {APP_PATH}")
if APP_PATH.parent != project_dir / "app":
    raise RuntimeError("app was imported from outside the intended checkout")
```

```python
print(f"Project root: {project_dir}")
print(f"Dotenv file: {env_dir}")
print(f"Dotenv file exists: {env_dir.is_file()}")
print(f"Raw data directory: {data_dir}")
print(f"Output root: {OUTPUT_ROOT}")
print(f"Source reference: {SOURCE_REFERENCE}")
print(f"Source reference exists: {SOURCE_REFERENCE.is_file()}")
print(f"Matplotlib config directory: {mplconfig_dir}")
```
